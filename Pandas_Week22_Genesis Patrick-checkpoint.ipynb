# Group by and Aggregating

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Genesis Patrick\Documents\DAM4Ace Files\Week 23\Flavors.csv")
df
```

```python
#running aggregations to get an output
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
#running aggregations to get an output by Mean
```

```python
df.groupby('Base Flavor').mean(numeric_only=True)
```

```python
#running aggregations to get an output by Count
```

```python
df.groupby('Base Flavor').count()
```

```python
#running aggregations to get an output by Min
```

```python
df.groupby('Base Flavor').min()
```

```python
#running aggregations to get an output by Max
```

```python
df.groupby('Base Flavor').max()
```

```python
#running aggregations to get an output by Sum
```

```python
df.groupby('Base Flavor').sum()
```

```python
#running aggregate function
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean','min','max','count','sum']})
```

```python
df.groupby('Base Flavor').agg({'Texture Rating': ['mean','min','max','count','sum']})
```

```python
#Grouping on multiple columns
```

```python
df
```

```python
df.groupby(['Base Flavor','Liked']).mean(numeric_only=True)
```

```python
#running aggregate function
```

```python
df.groupby(['Base Flavor','Liked']).agg({'Flavor Rating': ['mean','min','max','count','sum']})
```

```python
#Shortcut Finction for Describe
```

```python
df.groupby('Base Flavor').describe()
```
